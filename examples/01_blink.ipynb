# Blink

Run `python -m nanopynq doctor` first. If the board does not respond, unplug and re-plug the USB cable, then run the `Overlay(...)` cell again (it re-programs the board, then opens the port). Use `download=False` only to reopen within the same power cycle.

```python
from nanopynq import Overlay

ol = Overlay("base", board="tangnano9k")
```

```python
ol.leds[0].on()
```

```python
ol.buttons[0].read()
```

```python
import time

for _ in range(5):
    ol.leds[0].on()
    time.sleep(0.3)
    ol.leds[0].off()
    time.sleep(0.3)
```
